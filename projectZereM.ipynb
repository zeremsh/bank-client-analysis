# Изменение типов данных

```python
# Initialize new purpose column
bank_clients['new_purpose'] = None

# Function to categorize based on substring matching
def categorize_purpose(purpose):
    purpose_lower = purpose.lower()
    
    if any(x in purpose_lower for x in ['свадьб', 'сыг']): 
        return 'Проведение свадьбы'
    elif any(x in purpose_lower for x in ['образова']): 
        return 'Получение образования'  
    elif any(x in purpose_lower for x in ['жил', 'кварти', 'ремон']): 
        return 'Покупка квартиры/жилья'
    elif any(x in purpose_lower for x in ['автомо', 'машин']): 
        return 'Приобретение автомобиля'
    elif any(x in purpose_lower for x in ['недвижим']): # недвижимость, строительство, коммерческая
        return 'Покупка недвижимости'
    else:
        return None

# Apply categorization to each row
bank_clients['new_purpose'] = bank_clients['purpose'].apply(categorize_purpose)
```

# Корректировка ошибок

```python
bank_clients.describe()
```

```python
# Округление значения в столбце ежемесячного дохода
bank_clients['total_income'] = bank_clients['total_income'].round(1)
# Округление значений в столбце стажа работы и возведение по модулю для вывода положительных значений
bank_clients['days_employed'] = bank_clients['days_employed'].round(0).abs()
# Возведение по модулю для положительных значений в столбце с детьми
bank_clients['children'] = bank_clients['children'].abs()
```

```python
# Предположим, что 20 детей это ошибка и 0 был поставлен случайно. Тогда меняем 20 на 2:
bank_clients.loc[bank_clients['children'] == 20, 'children'] = 2
```

```python
# Создаем таблицу для анализа взаимосвязи между семейным положением и долгом
family_debt_crosstab = pd.crosstab(bank_clients['family_status'], bank_clients['debt'])

# Расчет % доли для каждого семейного положения
family_debt_pct = pd.crosstab(bank_clients['family_status'], bank_clients['debt'], normalize='index') * 100

# Отображение необработанных значении
print("Raw counts of debt by family status:")
print(family_debt_crosstab)
print("\nPercentage of debt within each family status:")
print(family_debt_pct.round(2))

# Визуализация значении
plt.figure(figsize=(10, 6))
family_debt_pct[1].plot(kind='bar')
plt.title('Percentage of Debt by Family Status')
plt.xlabel('Family Status')
plt.ylabel('Percentage with Debt')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Есть ли зависимость между уровнем дохода и возвратом кредита в срок?

```python
# Расчет % доли ежемесячного дохода
income_33 = bank_clients['total_income'].quantile(0.33)
income_66 = bank_clients['total_income'].quantile(0.66)

# Создание категории дохода на основе % доли
def categorize_income(income):
    if income <= income_33:
        return 'Low income'
    elif income <= income_66:
        return 'Medium income'
    else:
        return 'High income'

# Добавляем новый столбец income_category
bank_clients['income_category'] = bank_clients['total_income'].apply(categorize_income)
```

```python
# Расчет среднего значения по долгам в income_category
defaults = bank_clients.groupby('income_category')['debt'].mean()

# Создаем гистограмму
plt.figure(figsize=(10, 6))
plt.bar(defaults.index, defaults)
plt.title('Debt Default Rate by Income Category')
plt.xlabel('Income Category') 
plt.ylabel('Default Rate')

plt.show()


```

#### Как разные цели кредита влияют на его возврат в срок?


```python
# Расчет дефолтного значения по столбцу цели кредита
defaults_by_purpose = bank_clients.groupby('new_purpose')['debt'].mean() * 100

# Создание гистограммы
plt.figure(figsize=(10, 6))
plt.bar(defaults_by_purpose.index, defaults_by_purpose)
plt.title('Default Rate by Credit Purpose')
plt.xlabel('Credit Purpose')
plt.ylabel('Default Rate (%)')

# Коррекция поворота х-оси для лучшей читаемости
plt.xticks(rotation=45, ha='right')

# Добавление значений сверху каждой колоны
for i, v in enumerate(defaults_by_purpose):
    plt.text(i, v + 0.5, f'{v:.1f}%', ha='center')

# Отрегулирование колон, чтобы он не был урезан
plt.tight_layout()


plt.show()

# Ввод нумер значении
print("\nDefault rates by credit purpose:")
print(defaults_by_purpose.sort_values(ascending=False))
```
